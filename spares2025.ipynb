```python
# Import Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

## Data Understanding

```python
# Import the data to the notebook
df1= pd.read_excel('/content/drive/MyDrive/Vic/Spare-parts/SPARES 2025/BUTALI HQ.xlsx', skiprows=6)
df2= pd.read_excel('/content/drive/MyDrive/Vic/Spare-parts/SPARES 2025/CHEKALINI.xlsx', skiprows=6)
df3= pd.read_excel('/content/drive/MyDrive/Vic/Spare-parts/SPARES 2025/MATILI.xlsx', skiprows=5)
df4= pd.read_excel('/content/drive/MyDrive/Vic/Spare-parts/SPARES 2025/TURBO.xlsx', skiprows=7)
```

```python
# Combine all the datasets into one
df= pd.concat([df1, df2, df3, df4], ignore_index=True)
df.head()
```

```python
# Check the shape of the dataset
df.shape
```

```python
# General info of the dataset
df.info()
```

## Data Cleaning

```python
# Drop all the columns that are unnamed
df = df.loc[:, ~df.columns.str.startswith('Unnamed')]
df.columns
```

```python
# There are also columns with no information, drop them
# Drop the initial index
df= df.drop(columns=['Req. No', 'Job No', 'No..', 'Pump Number'])
df.columns
```

```python
# Check for null values
df.isnull().sum()
```

```python
df[df.isnull().any(axis=1)]
```

```python
#Check the rows missing in unit price column
df[df['Unit Price'].isnull()]
```

```python
# Drop the rows missing in unit price column
df= df.dropna(subset='Unit Price')
df.info()
```

```python
#Check the rows missing in unit price column
df[df['Issued To'].isnull()]
```

```python
# Fill in the missing value with "Workshop Use"
df['Issued To']= df['Issued To'].fillna('Workshop Use')
```

```python
# Replace missing values in driver column with "Unknown"
df['Driver']=df['Driver'].fillna('Unknown')
```

```python
# Check for any more missing values
df.isnull().sum()
```

```python
# Convert all capital letters in column names to small
# Replace spaces with underscore
df.columns = df.columns.str.lower().str.replace(' ', '_')
```

```python
# Rename columns with extra full stops
df=df.rename(columns={'issue_no..':'issue_no','date..':'date'})
```

## Exploratory Data Analysis


```python
#Check the remaining columns and rows
df.info()
```

```python
# Check the general analysis of the numerical data
df.describe()
```

```python
# Check store that issues the most spares
plt.figure(figsize=(10,6))
ax= df.groupby('store')['quantity'].sum().sort_values(ascending=False).plot(kind='bar')
ax.bar_label(ax.containers[0]);
```

Butali workshop store has the highestb number of issuance in regards to spare parts. This can be tied to the fact that it has the highest number of units, followed by Turbo statio, then Chekalini station and lastly Matili

```python
# Check frequency of drivers to the workshop
plt.figure(figsize=(10,6))
ax=df[df['driver'] != 'Unknown']['driver'].value_counts().head(10).plot(kind='barh')
ax.bar_label(ax.containers[0]);
```

```python
# Check item frequency
plt.figure(figsize=(10,6))
ax=df['item_desc'].value_counts().head(10).plot(kind='barh')
ax.bar_label(ax.containers[0]);
```

```python
# check for relationship between months and cost(value)
plt.figure(figsize=(10,6))
df['month'] = df['date'].dt.strftime('%b')  # Jan, Feb, Mar...
plt.scatter(df['month'], df['value']);
```

```python
# group drivers by exependiture
plt.figure(figsize=(10,6))
ax=df.groupby('driver')['value'].sum().sort_values(ascending=False).iloc[1:11].plot(kind='barh')
ax.bar_label(ax.containers[0]);
```

```python
# Check expenditure across the months.
plt.figure(figsize=(10,6))
df.groupby(df['date'].dt.month_name())['value'].sum().reindex(
    ['January','February','March','April','May','June','July',
     'August','September','October','November','December']
).plot(kind='bar')
```

```python
#Check to see which months had the highest expenditure
plt.figure(figsize=(10,6))
df.groupby(df['date'].dt.month_name())['value'].sum().sort_values(ascending=False).plot(kind='bar')
```

```python
# line Graph for spare consumption
monthly_issues = df.groupby('month')['value'].sum()

monthly_issues.plot(kind='line', figsize=(10, 6), marker='o')
```

```python
df['month_number'] = df['date'].dt.strftime('%m')
station_consumption = df.groupby(['month_number', 'store'])['quantity'].sum().unstack()

station_consumption.plot(
    kind='line',
    figsize=(12, 6),
    marker='o'
)

plt.title('Spare Parts Consumption by Station')
plt.xlabel('Month')
plt.ylabel('Quantity Consumed')
plt.legend(title='Station')
plt.xticks(rotation=45)
plt.show()
```

```python

df['month_number'] = df['date'].dt.strftime('%m')
monthly_value = df.groupby(['month_number','store'])['value'].sum().unstack()

monthly_value.plot(
    kind='line',
    figsize=(10, 6),
    marker='o'
)
```
